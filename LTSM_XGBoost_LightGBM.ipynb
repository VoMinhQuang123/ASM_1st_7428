```python
from prophet import Prophet


def sales_prediction(Store_ID, sales_df, holidays, periods):
    # Function that takes in the storeID and returns two date/sales columns in Prophet format
    # Format data to fit prophet

    sales_df = sales_df[sales_df['Store'] == Store_ID]
    print(f"Store #{Store_ID} has {len(sales_df)} records")
    sales_df = sales_df[['Date', 'Sales', 'Promo', 'CompetitionOpen']].rename(columns={'Date': 'ds', 'Sales': 'y'})
    sales_df = sales_df.sort_values('ds')

    model = Prophet(yearly_seasonality=True, weekly_seasonality=True, holidays=holidays, interval_width=0.95)
    model.add_regressor('Promo')
    model.add_regressor('CompetitionOpen')
    model.fit(sales_df)
    future = model.make_future_dataframe(periods=periods)
    future['Promo'] = sales_df['Promo'].median()
    future['CompetitionOpen'] = sales_df['CompetitionOpen'].median()

    forecast = model.predict(future)
    figure = model.plot(forecast, xlabel='Date', ylabel='Sales')
    figure2 = model.plot_components(forecast)
    return forecast.tail(42)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

def sales_prediction_lstm(Store_ID, sales_df, periods=42):
    # 1. Lọc dữ liệu cửa hàng
    sales_df = sales_df[sales_df['Store'] == Store_ID].copy()
    sales_df = sales_df.sort_values('Date')
    print(f"Store #{Store_ID} has {len(sales_df)} records")

    # 2. Feature Engineering
    sales_df['Date'] = pd.to_datetime(sales_df['Date'])
    sales_df['day'] = sales_df['Date'].dt.day
    sales_df['month'] = sales_df['Date'].dt.month
    sales_df['weekday'] = sales_df['Date'].dt.weekday
    sales_df['year'] = sales_df['Date'].dt.year
    sales_df['StateHoliday'] = sales_df['StateHoliday'].apply(lambda x: 0 if x == '0' else 1)

    sales_df['lag_1'] = sales_df['Sales'].shift(1)
    sales_df['lag_7'] = sales_df['Sales'].shift(7)
    sales_df['rolling_mean_7'] = sales_df['Sales'].shift(1).rolling(window=7).mean()

    sales_df.dropna(inplace=True)

    # 3. Lấy feature + scale
    features = ['day', 'month', 'weekday', 'year', 'Promo', 'CompetitionOpen', 'StateHoliday', 'lag_1', 'lag_7', 'rolling_mean_7']
    target = ['Sales']

    scaler_X = MinMaxScaler()
    scaler_y = MinMaxScaler()

    X_scaled = scaler_X.fit_transform(sales_df[features])
    y_scaled = scaler_y.fit_transform(sales_df[target])

    # 4. Reshape cho LSTM: (samples, timesteps, features)
    X_lstm = []
    y_lstm = []

    timesteps = 7  # số ngày dùng để dự đoán
    for i in range(timesteps, len(X_scaled)):
        X_lstm.append(X_scaled[i - timesteps:i])
        y_lstm.append(y_scaled[i])

    X_lstm, y_lstm = np.array(X_lstm), np.array(y_lstm)

    # 5. Train/Test split
    X_train, X_test = X_lstm[:-periods], X_lstm[-periods:]
    y_train, y_test = y_lstm[:-periods], y_lstm[-periods:]

    # 6. Xây dựng mô hình LSTM
    model = Sequential()
    model.add(LSTM(64, activation='relu', input_shape=(X_train.shape[1], X_train.shape[2])))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse')

    # 7. Train
    model.fit(X_train, y_train, epochs=30, batch_size=16, verbose=0)

    # 8. Dự đoán
    y_pred_scaled = model.predict(X_test)
    y_pred = scaler_y.inverse_transform(y_pred_scaled)
    y_true = scaler_y.inverse_transform(y_test)

    # 9. Vẽ biểu đồ
    plt.figure(figsize=(12,6))
    plt.plot(sales_df.iloc[-periods:]['Date'], y_true, label='True Sales', color='blue')
    plt.plot(sales_df.iloc[-periods:]['Date'], y_pred, label='LSTM Predicted Sales', color='green')
    plt.title(f'Store #{Store_ID} Sales Forecast (LSTM)')
    plt.xlabel('Date')
    plt.ylabel('Sales')
    plt.legend()
    plt.grid(True)
    plt.show()

    # 10. RMSE
    from sklearn.metrics import mean_squared_error
    print("RMSE:", np.sqrt(mean_squared_error(y_true, y_pred)))

    # 11. Kết quả
    forecast_df = pd.DataFrame({
        'Date': sales_df.iloc[-periods:]['Date'].values,
        'True Sales': y_true.flatten(),
        'Predicted Sales': y_pred.flatten()
    })

    return forecast_df


```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

def forecast_sales_lstm_future(Store_ID, sales_df, periods=42):
    # 1. Lọc dữ liệu cửa hàng
    sales_df = sales_df[sales_df['Store'] == Store_ID].copy()
    sales_df = sales_df.sort_values('Date')
    sales_df['Date'] = pd.to_datetime(sales_df['Date'])

    # 2. Feature engineering
    sales_df['day'] = sales_df['Date'].dt.day
    sales_df['month'] = sales_df['Date'].dt.month
    sales_df['weekday'] = sales_df['Date'].dt.weekday
    sales_df['year'] = sales_df['Date'].dt.year
    sales_df['StateHoliday'] = sales_df['StateHoliday'].apply(lambda x: 0 if x == '0' else 1)

    sales_df['lag_1'] = sales_df['Sales'].shift(1)
    sales_df['lag_7'] = sales_df['Sales'].shift(7)
    sales_df['rolling_mean_7'] = sales_df['Sales'].shift(1).rolling(window=7).mean()

    sales_df.dropna(inplace=True)

    features = ['day', 'month', 'weekday', 'year', 'Promo', 'CompetitionOpen', 'StateHoliday', 'lag_1', 'lag_7', 'rolling_mean_7']
    target = ['Sales']

    # 3. Scale dữ liệu
    scaler_X = MinMaxScaler()
    scaler_y = MinMaxScaler()

    X_scaled = scaler_X.fit_transform(sales_df[features])
    y_scaled = scaler_y.fit_transform(sales_df[target])

    timesteps = 7
    X_lstm = []
    y_lstm = []

    for i in range(timesteps, len(X_scaled)):
        X_lstm.append(X_scaled[i - timesteps:i])
        y_lstm.append(y_scaled[i])

    X_lstm, y_lstm = np.array(X_lstm), np.array(y_lstm)

    # 4. Train mô hình
    model = Sequential()
    model.add(LSTM(64, activation='relu', input_shape=(X_lstm.shape[1], X_lstm.shape[2])))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse')
    model.fit(X_lstm, y_lstm, epochs=30, batch_size=16, verbose=0)

    # 5. Dự đoán 42 ngày tương lai
    last_known_date = sales_df['Date'].max()
    future_dates = [last_known_date + pd.Timedelta(days=i) for i in range(1, periods + 1)]

    last_sequence = X_scaled[-timesteps:]  # shape: (7, n_features)
    forecast_sales = []

    for i in range(periods):
        input_seq = last_sequence.reshape(1, timesteps, X_scaled.shape[1])
        pred_scaled = model.predict(input_seq, verbose=0)
        pred = scaler_y.inverse_transform(pred_scaled)[0][0]
        forecast_sales.append(pred)

        # Giả lập đặc trưng của ngày tiếp theo
        next_date = future_dates[i]
        next_row = {
            'day': next_date.day,
            'month': next_date.month,
            'weekday': next_date.weekday(),
            'year': next_date.year,
            'Promo': 0,  # bạn có thể thay đổi nếu biết có khuyến mãi
            'CompetitionOpen': sales_df['CompetitionOpen'].mean(),
            'StateHoliday': 0,
            'lag_1': pred,
            'lag_7': forecast_sales[i - 6] if i >= 6 else pred,
            'rolling_mean_7': np.mean(forecast_sales[-7:]) if len(forecast_sales) >= 7 else pred
        }

        next_row_scaled = scaler_X.transform(pd.DataFrame([next_row]))
        last_sequence = np.vstack([last_sequence[1:], next_row_scaled])  # cập nhật chuỗi

    # 6. Trả kết quả
    forecast_df = pd.DataFrame({
        'Date': future_dates,
        'Predicted Sales': forecast_sales
    })

    # 7. Vẽ
    plt.figure(figsize=(12,6))
    plt.plot(forecast_df['Date'], forecast_df['Predicted Sales'], label='Forecasted Sales (Next 42 Days)', color='green')
    plt.title(f'Store #{Store_ID} - Future 42-Day Sales Forecast (LSTM)')
    plt.xlabel('Date')
    plt.ylabel('Sales')
    plt.grid(True)
    plt.legend()
    plt.show()

    return forecast_df
```

```python
import pandas as pd
import numpy as np
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

def sales_prediction_xgb(Store_ID, sales_df, periods=42):
    # 1. Lọc dữ liệu của cửa hàng cụ thể
    sales_df = sales_df[sales_df['Store'] == Store_ID].copy()
    sales_df = sales_df.sort_values('Date')
    print(f"Store #{Store_ID} has {len(sales_df)} records")

    # 2. Feature Engineering
    sales_df['Date'] = pd.to_datetime(sales_df['Date'])
    sales_df['day'] = sales_df['Date'].dt.day
    sales_df['month'] = sales_df['Date'].dt.month
    sales_df['weekday'] = sales_df['Date'].dt.weekday
    sales_df['year'] = sales_df['Date'].dt.year

    # Optional: Chuyển StateHoliday thành nhị phân
    sales_df['StateHoliday'] = sales_df['StateHoliday'].apply(lambda x: 0 if x == '0' else 1)

    # Lag feature (để có yếu tố thời gian)
    sales_df['lag_1'] = sales_df['Sales'].shift(1)
    sales_df['lag_7'] = sales_df['Sales'].shift(7)
    sales_df['rolling_mean_7'] = sales_df['Sales'].shift(1).rolling(window=7).mean()

    sales_df.dropna(inplace=True)

    # 3. Tách train/test
    features = ['day', 'month', 'weekday', 'year', 'Promo', 'CompetitionOpen', 'StateHoliday', 'lag_1', 'lag_7', 'rolling_mean_7']
    X = sales_df[features]
    y = sales_df['Sales']

    X_train, X_test = X[:-periods], X[-periods:]
    y_train, y_test = y[:-periods], y[-periods:]

    # 4. Train XGBoost
    model = XGBRegressor(n_estimators=100, learning_rate=0.1)
    model.fit(X_train, y_train)

    # 5. Predict
    y_pred = model.predict(X_test)

    # 6. Plot
    plt.figure(figsize=(12,6))
    plt.plot(y_test.index, y_test, label='True Sales', color='blue')
    plt.plot(y_test.index, y_pred, label='XGBoost Predicted Sales', color='red')
    plt.title(f'Store #{Store_ID} Sales Forecast (XGBoost)')
    plt.xlabel('Time')
    plt.ylabel('Sales')
    plt.legend()
    plt.grid(True)
    plt.show()

    print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))

    # 7. Trả kết quả dự đoán
    forecast_df = pd.DataFrame({'Date': sales_df.iloc[-periods:]['Date'],
                                'True Sales': y_test.values,
                                'Predicted Sales': y_pred})
    return forecast_df
```

```python
import pandas as pd
import numpy as np
from xgboost import XGBRegressor
import matplotlib.pyplot as plt

def sales_prediction_xgb_forecast(Store_ID, sales_df, forecast_days=42):
    # 1. Lọc dữ liệu của cửa hàng
    sales_df = sales_df[sales_df['Store'] == Store_ID].copy()
    sales_df = sales_df.sort_values('Date')
    print(f"Store #{Store_ID} has {len(sales_df)} records")

    # 2. Feature Engineering
    sales_df['Date'] = pd.to_datetime(sales_df['Date'])
    sales_df['day'] = sales_df['Date'].dt.day
    sales_df['month'] = sales_df['Date'].dt.month
    sales_df['weekday'] = sales_df['Date'].dt.weekday
    sales_df['year'] = sales_df['Date'].dt.year
    sales_df['StateHoliday'] = sales_df['StateHoliday'].apply(lambda x: 0 if x == '0' else 1)

    sales_df['lag_1'] = sales_df['Sales'].shift(1)
    sales_df['lag_7'] = sales_df['Sales'].shift(7)
    sales_df['rolling_mean_7'] = sales_df['Sales'].shift(1).rolling(window=7).mean()
    sales_df.dropna(inplace=True)

    # 3. Train mô hình XGBoost
    features = ['day', 'month', 'weekday', 'year', 'Promo', 'CompetitionOpen', 'StateHoliday', 'lag_1', 'lag_7', 'rolling_mean_7']
    X = sales_df[features]
    y = sales_df['Sales']
    model = XGBRegressor(n_estimators=100, learning_rate=0.1)
    model.fit(X, y)

    # 4. Dự đoán 42 ngày trong tương lai
    future_data = []
    last_date = sales_df['Date'].max()
    prev_data = sales_df.copy()

    for i in range(forecast_days):
        date = last_date + pd.Timedelta(days=1)
        day = date.day
        month = date.month
        weekday = date.weekday()
        year = date.year

        # Giả định giá trị các feature khác không đổi (hoặc gán trung bình/gần nhất)
        Promo = prev_data['Promo'].iloc[-1]
        CompetitionOpen = prev_data['CompetitionOpen'].iloc[-1]
        StateHoliday = 0  # giả định không nghỉ lễ

        lag_1 = prev_data['Sales'].iloc[-1]
        lag_7 = prev_data['Sales'].iloc[-7] if len(prev_data) >= 7 else lag_1
        rolling_mean_7 = prev_data['Sales'].iloc[-7:].mean()

        features_row = [day, month, weekday, year, Promo, CompetitionOpen, StateHoliday, lag_1, lag_7, rolling_mean_7]
        pred_sales = model.predict([features_row])[0]

        # Lưu lại để làm lag cho ngày tiếp theo
        new_row = pd.DataFrame([{'Date': date, 'Sales': pred_sales}])
        prev_data = pd.concat([prev_data, new_row], ignore_index=True)
        future_data.append({'Date': date, 'Predicted Sales': pred_sales})

        last_date = date

    # 5. Vẽ biểu đồ
    forecast_df = pd.DataFrame(future_data)
    plt.figure(figsize=(12,6))
    plt.plot(forecast_df['Date'], forecast_df['Predicted Sales'], label='Forecasted Sales (Next 42 Days)', color='green')
    plt.title(f'Store #{Store_ID} - Future 42-Day Sales Forecast (XGBoost)')
    plt.xlabel('Date')
    plt.ylabel('Sales')
    plt.legend()
    plt.grid(True)
    plt.show()

    return forecast_df



```

```python
import pandas as pd
import numpy as np
from lightgbm import LGBMRegressor
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

def sales_prediction_lgbm(Store_ID, sales_df, periods=42):
    # 1. Lọc dữ liệu của cửa hàng cụ thể
    sales_df = sales_df[sales_df['Store'] == Store_ID].copy()
    sales_df = sales_df.sort_values('Date')
    print(f"Store #{Store_ID} has {len(sales_df)} records")

    # 2. Feature Engineering
    sales_df['Date'] = pd.to_datetime(sales_df['Date'])
    sales_df['day'] = sales_df['Date'].dt.day
    sales_df['month'] = sales_df['Date'].dt.month
    sales_df['weekday'] = sales_df['Date'].dt.weekday
    sales_df['year'] = sales_df['Date'].dt.year

    # Optional: Chuyển StateHoliday thành nhị phân
    sales_df['StateHoliday'] = sales_df['StateHoliday'].apply(lambda x: 0 if x == '0' else 1)

    # Lag feature (để có yếu tố thời gian)
    sales_df['lag_1'] = sales_df['Sales'].shift(1)
    sales_df['lag_7'] = sales_df['Sales'].shift(7)
    sales_df['rolling_mean_7'] = sales_df['Sales'].shift(1).rolling(window=7).mean()

    sales_df.dropna(inplace=True)

    # 3. Tách train/test
    features = ['day', 'month', 'weekday', 'year', 'Promo', 'CompetitionOpen',
                'StateHoliday', 'lag_1', 'lag_7', 'rolling_mean_7']
    X = sales_df[features]
    y = sales_df['Sales']

    X_train, X_test = X[:-periods], X[-periods:]
    y_train, y_test = y[:-periods], y[-periods:]

    # 4. Train LightGBM
    model = LGBMRegressor(n_estimators=100, learning_rate=0.1)
    model.fit(X_train, y_train)

    # 5. Predict
    y_pred = model.predict(X_test)

    # 6. Plot
    plt.figure(figsize=(12,6))
    plt.plot(y_test.index, y_test, label='True Sales', color='blue')
    plt.plot(y_test.index, y_pred, label='LightGBM Predicted Sales', color='green')
    plt.title(f'Store #{Store_ID} Sales Forecast (LightGBM)')
    plt.xlabel('Time')
    plt.ylabel('Sales')
    plt.legend()
    plt.grid(True)
    plt.show()

    print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))

    # 7. Trả kết quả dự đoán
    forecast_df = pd.DataFrame({
        'Date': sales_df.iloc[-periods:]['Date'],
        'True Sales': y_test.values,
        'Predicted Sales': y_pred
    })

    return forecast_df
```
